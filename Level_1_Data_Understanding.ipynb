```python
df_sorted=df.sort_values(["Train_No", "Route_Number"])
```

```python
train_stations = df_sorted.groupby("Train_No").agg(
    Starting_Station=("Station_Name", "first"),
    Ending_Station=("Station_Name", "last")
).reset_index()

train_stations.head()
```

```python
train_distance= df_sorted.groupby("Train_No")["Distance"].max()
train_distance.head()
```

```python
train_stops=df_sorted.groupby("Train_No").size()
train_stops.head()
```
